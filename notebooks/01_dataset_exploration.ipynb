# IIIT-5K Dataset Exploration & Analysis

This notebook explores the **IIIT-5K Words** dataset used for training and evaluating our Optical Character Recognition (OCR) and Text Redaction pipeline.

### Objectives:
1. Inspect dataset partitions (`traindata.csv`, `testdata.csv`).
2. Analyze character set vocabulary and label length distributions.
3. Inspect character-level bounding box annotations (`trainCharBound.csv`, `testCharBound.csv`).
4. Inspect image properties (resolutions, aspect ratios, channels).
5. Demonstrate word-level bounding box construction.

```python
import os
import cv2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

DATA_DIR = os.path.join("..", "IIIT5K-Word_V3.0", "IIIT5K")
if not os.path.exists(DATA_DIR):
    DATA_DIR = os.path.join("IIIT5K-Word_V3.0", "IIIT5K")

train_df = pd.read_csv("../traindata.csv" if os.path.exists("../traindata.csv") else "traindata.csv")
test_df = pd.read_csv("../testdata.csv" if os.path.exists("../testdata.csv") else "testdata.csv")

print(f"Train samples: {len(train_df)}")
print(f"Test samples: {len(test_df)}")
train_df.head()
```

### Character Set Vocabulary Analysis
The dataset consists of cropped scene text images representing alphanumeric words.

```python
all_chars = sorted(list(set(''.join(train_df['GroundTruth'].dropna().astype(str).str.upper()))))
print(f"Vocabulary size ({len(all_chars)} alphanumeric characters):")
print(all_chars)

# Length distribution
lengths = train_df['GroundTruth'].astype(str).str.len()
print(f"Min word length: {lengths.min()}, Max: {lengths.max()}, Mean: {lengths.mean():.2f}")
```

### Inspecting Character Bounding Boxes

```python
char_file = "../trainCharBound.csv" if os.path.exists("../trainCharBound.csv") else "trainCharBound.csv"
char_df = pd.read_csv(char_file)
print(char_df.head(2))

# Sample visualization of word image
sample_row = train_df.iloc[0]
sample_path = os.path.join(DATA_DIR, sample_row['ImgName'])
img = cv2.imread(sample_path)
print(f"Image: {sample_row['ImgName']}, Label: {sample_row['GroundTruth']}, Shape: {img.shape}")
```
